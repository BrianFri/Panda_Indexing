### Task 8: Select Rows With .iloc

```python
print("Position 0: \n", ag_df.iloc[0])
print("\nLast Position:\n", ag_df.iloc[-1])
first_five = ag_df.iloc[0:5]
print(first_five)
```

1. iloc[0:5] -- how many rows did you get? Why not 6?
5 row because it is not inclusive.
2. What is the difference between .iloc[0:5] and .loc[0:5]?
.loc will give you 6 rows while .iloc will only give you 5 rows.
